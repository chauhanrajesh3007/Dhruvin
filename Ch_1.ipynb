# merge(left,right,how,on,left_on,rigt_on,left_index,right_index,sort,suffixes=('_x','_y')

```python
import pandas as pd 
df1=pd.DataFrame({'Rno':[10,20,30,40],
                 "Name":['A',"B",'C','E'],
                 "SPI":[7,8,7.2,9.2]})

df2=pd.DataFrame({'Rno':[10,20,50],
                 "Name":['A',"B",'D'],
                 "CPI":[7,8,9.2]})
```

```python
res=pd.merge(left=df1,right=df2)
print(res)
```

```python
res=pd.merge(left=df1,right=df2,how="outer")
print(res)
```

```python
import pandas as pd 
df1=pd.DataFrame({'Rno':[10,20,30,40],
                 "Name":['A',"B",'C','E'],
                 "SPI":[7,8,7.2,5.2]})

df2=pd.DataFrame({'Rno':[20,50,10],
                 "Name":['AA',"BB",'DD'],
                 "CPI":[7.5,8.9,9.2]})
```

```python
res=pd.merge(left=df1,right=df2,how="outer",on='Rno')
print(res)
```

```python
import pandas as pd 
df1=pd.DataFrame({'Rno':[10,20,30,40],
                 "Name":['A',"B",'C','E'],
                 "SPI":[7,8,7.2,5.2]})

df2=pd.DataFrame({'Eno':[20,50,10],
                 "Name":['AA',"BB",'DD'],
                 "CPI":[7.5,8.9,9.2]})
```

```python
res=pd.merge(left=df1,right=df2,how="outer",left_on="Rno",right_on="Eno")
print(res)
```

```python
res=pd.merge(left=df1,right=df2,how="inner",left_on="Rno",right_on="Eno",suffixes=('_df1','_df2'))
print(res)
```

```python
res=pd.merge(left=df1,right=df2,how="left",left_on="Rno",right_on="Eno",suffixes=('_df1','_df2'))
print(res)
```

```python
res=pd.merge(left=df1,right=df2,how="inner",left_index=True,right_index=True,suffixes=('_df1','_df2'))
print(res)
```

```python
res=pd.merge(left=df1,right=df2,how="outer",left_index=True,right_index=True,suffixes=('_df1','_df2'))
print(res)
```

```python
import pandas as pd
df1=pd.DataFrame({'branch':['IT','IT','CE','ML','CE'],
                 'SPI':[1,2,3,4,7],
                 'CPI':[1.2,2.2,3.2,5.2,7.2]})
print(df1)
```

```python
res=df1.groupby(by='branch')['SPI'].sum()
print(res)
```

```python
res=df1.groupby(by='branch',as_index=False,sort=False)['SPI'].sum()
print(res)
```

```python
res=df1.groupby(by='branch',as_index=False,sort=True)['SPI'].sum()
print(res)
```

```python
res=df1.groupby(by='branch')['SPI'].mean()
print(res)
```
