## get_dummies(data,columns,prefix,drop_first)

```python
import pandas as pd
s=pd.Series(['C','A','D','B'])
res = pd.get_dummies(s)
print(res)
```

```python
import pandas as pd
s=pd.Series(['C','A','D','B'])
res = pd.get_dummies(s,drop_first=True,prefix='Cat')
print(res)
```

```python
import pandas as pd
df=pd.DataFrame({'Area' : [500,750,1000,1200,1500],
               'Price':['Low','Low','Mid','High','High']})
res =pd.get_dummies(df)
print(res)
```

```python
import pandas as  pd
df=pd.DataFrame({'Area':[500,750,1000,1200,1500],
                'Price':['Low','Low','Mid','High','High'],
                'category':['A','B','A','B','A']})
res=pd.get_dummies(df,prefix='AA')
print(res)
```

```python
import pandas as  pd
df=pd.DataFrame({'Area':[500,750,1000,1200,1500],
                'Price':['Low','Low','Mid','High','High'],
                'category':['A','B','A','B','A']})
res=pd.get_dummies(df)
print(res)
```

```python
import pandas as  pd
df=pd.DataFrame({'Area':[500,750,1000,1200,1500],
                'Price':['Low','Low','Mid','High','High'],
                'category':['A','B','A','B','A']})
res=pd.get_dummies(df,columns=['Price'])
print(res)
```

# np.where()


```python
import numpy as np
import pandas as pd
age=np.array(range(10,65))
df=pd.DataFrame({'Age':age})

df['Age_Group']=np.where(df['Age']<=20,'Teen','Adult')
print(df)
```

```python
import numpy as np
import pandas as pd
age=np.array(range(10,65))
df=pd.DataFrame({'Age':age})

df['Age_Group']=np.where(df['Age']<=20,'Teen',np.where(df['Age']<=35,'Young',np.where(df['Age']<=50,'mid','Senior')))
print(df)
```

```python
import pandas as  pd
df=pd.DataFrame({'Area':[500,750,1000,1200,1500],
                'Price':[1500000,1800000,2200000,3100000,3500000]})
df['Category']=np.where(df['Price']>3000000,'High',
               np.where(df['Price']<2000000,'Low',
                        'Mid'))
print(df)
```

```python
import numpy as np
import pandas as pd
age=np.array(range(10,65))
df=pd.DataFrame({'Age':age})

df['Age_Group']=np.where(df['Age']<30,'young',np.where(df['Age']<=60, 'Middle-aged','Elderly'))
print(df)
```
