# Concat(obj, axis, join, ignore_index, sort)

```python
import pandas as pd
import numpy as np
df1=pd.DataFrame(
    {
    'Name':['Aaryan','Bob','Carla'],
    'Age':[20,22,23],
    'Branch':['IT','AI','CSE'],
    }
)
df2=pd.DataFrame(
    {
    'Name':['Aaryan','Bob','Carla','Bob3','Carla3','Bob2'],
    'Age':[20,22,23,25,20,22],
    'Branch':['IT','AI','CSE','AI','ML','AI'],
    }
)
```

```python
ndf=pd.concat([df1,df2],axis=0,join='outer') #ROW Marge
print(ndf)
```

```python
ndf=pd.concat([df1,df2],axis=1,join='outer') #Column Marge
print(ndf)
```

```python
ndf=pd.concat([df1,df2],axis=0,join='outer',ignore_index=True) #Index Change
print(ndf)
```

# merge(left,right,how='outer',on,left_on,right_on,left_index,right_index,sort,suffixes=('_x''_y'))
* on mean unique value

```python
import pandas as pd
import numpy as np
df1=pd.DataFrame(
    {
    'Name':['Aaryan','Bob','Carla'],
    'Rn':[1,2,3],
    'Branch':['IT','AI','CSE'],
    }
)
df2=pd.DataFrame(
    {
    'Name':['Aaryan','Bob','Carla','Bob3','Carla3','Bob2'],
    'Rn':[3,2,1,4,5,6],
    'C':['IT','AI','CSE','AI','ML','AI'],
    }
)
```

```python
res=pd.merge(left=df1,right=df2,how='outer',on='Rn')
print(res)
#Value base joining
```

```python
res=pd.merge(left=df1,right=df2,how='outer',on='Rn',suffixes=('_df1','_df2'))
print(res)
#Value base joining
# Column name same use on
```

```python
import pandas as pd
import numpy as np
df1=pd.DataFrame(
    {
    'Name':['Aaryan','Bob','Carla'],
    'Rn':[6,2,1],
    'Branch':['IT','AI','CSE'],
    }
)
df2=pd.DataFrame(
    {
    'Name':['Aaryan','Bob','Carla','Bob3','Carla3','Bob2'],
    'Rolln':[6,2,1,4,5,3],
    'C':['IT','AI','CSE','AI','ML','AI'],
    }
)
```

```python
res=pd.merge(left=df1,right=df2,how='outer',left_on='Rn',right_on='Rolln',suffixes=('_df1','_df2'))
print(res)
#Value base joining
# Column name not same use left_on,right_on
```

```python
res=pd.merge(left=df1,right=df2,how='inner',left_on='Rn',right_on='Rolln',suffixes=('_df1','_df2'),sort=False)
print(res)
#Value base joining
# Column name not same use left_on,right_on
```

```python
res=pd.merge(left=df1,right=df2,how='inner',left_on='Rn',right_on='Rolln',suffixes=('_df1','_df2'),sort=True)
print(res)
#Value base joining
#value wise sort
```

```python
import pandas as pd
import numpy as np
df1=pd.DataFrame(
    {
    'Name':['Aaryan','Bob','Carla'],
    'Rn':[6,2,10],
    'Branch':['IT','AI','CSE'],
    }
)
df2=pd.DataFrame(
    {
    'Name':['Aaryan','Bob','Carla','Bob3','Carla3','Bob2'],
    'Rolln':[6,2,1,4,5,3],
    'C':['IT','AI','CSE','AI','ML','AI'],
    }
)
```

```python
res=pd.merge(left=df1,right=df2,how='left',left_on='Rn',right_on='Rolln',suffixes=('_df1','_df2'))
print(res)
#Value base joining
```

```python
res=pd.merge(left=df1,right=df2,how='right',left_on='Rn',right_on='Rolln',suffixes=('_df1','_df2'))
print(res)
#Value base joining
```

```python
res=pd.merge(left=df1,right=df2,how='outer',left_index=True,right_index=True)
print(res)
#Value base joining
```

```python
df2=pd.DataFrame(
    {
    'SPI':[10,9,8,7,7,8,9],
    'Rolln':[1,2,3,4,5,6,7],
    'Branch':['IT','AI','CSE','AI','IT','AI','CSE'],
    }
)
print(df2)
res=df2.groupby(by='Branch')['SPI'].min()
print("------------------------------------------------\n")
print(res)
```

```python
df2=pd.DataFrame(
    {
    'SPI':[10,9,8,7,7,8,9,8],
    'Rolln':[1,2,3,4,5,6,7,10],
    'Branch':['IT','IT','AI','AI','AI','CSE','CSE','CSE'],
    }
)
print(df2)
res=df2.groupby(by='Branch',as_index=False)['SPI'].min()
print("------------------------------------------------\n")
print(res)
```

```python
print(df2)
res=df2.groupby(by='Branch',as_index=False,sort=False)['SPI'].min()
print("------------------------------------------------\n")
print(res)
#Alphabatically order
```

```python
print(df2)
res=df2.groupby(by='Branch',as_index=False)['SPI'].sum()
print("------------------------------------------------\n")
print(res)
#Alphabatically order
```

```python
print(df2)
res=df2.groupby(by='Branch',as_index=False)['SPI'].agg(['min','max','sum'])
print("------------------------------------------------\n")
print(res)
#Alphabatically order
```

```python
print(df2)
res=df2.groupby(by='Branch').sample()
print("------------------------------------------------\n")
print(res)
#Randam value
```

```python
print(df2)
res=df2.groupby(by='Branch').groups
print("------------------------------------------------\n")
print(res)
#Randam value
```

```python
print(df2)
res=df2.groupby(by='Branch').nth[-1]
print("------------------------------------------------\n")
print(res)
#Randam value
```

## 145) You have been hired as a network analyst by a company to analyze the social network of their employees. The company has
provided you with the following data:
 There are 5 employees in the company, each identified by a unique ID from 1 to 5.
 The following relationships exist between the employees:
 1. Employee 1 is friends with Employee 2 and Employee 3.
 2. Employee 2 is friends with Employee 4.
 3. Employee 3 is friends with Employee 5.
 Your task is to create a NetworkX graph representing this social network and display it

```python
import networkx as nx
g=nx.DiGraph()
g.add_nodes_from(["E1",'E2',"E3",'E4','E5'])
g.add_edges_from([('E1','E2'),('E1','E3'),('E2','E4'),('E3','E5')])
nx.draw(g, with_labels=True, node_size=500,node_color='m',width=2, edge_color='r')
```
