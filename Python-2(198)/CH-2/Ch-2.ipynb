# Detection and Remove

```python
import pandas as pd
df=pd.read_csv('auto-mpg.csv')
def out_detection(df,col):
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    IQR=Q3-Q1
    lower=Q1-(IQR*1.5)
    upper=Q3+(IQR*1.5)
    ans=df.loc[(df[col]<lower) | (df[col]>upper)]
    print(ans)

out_detection(df,'mpg')
```

```python
import plotly.express as px
import pandas as pd

data = {'Category': ['A',"B",'C','A',"B",'C'],
       'Value':[10,20,15,12,18,17],
       'Region':['Eats', 'West','Eats', 'West','Eats', 'West'],
       'Details':['Info1','Info2','Info3','Info4','Info5','Info6']}

df = pd.DataFrame(data)

fig=px.bar(df,
          x='Category',
          y='Value',
          title='Bar Chart',
          orientation='v',
          height=400,
          width=600,
          barmode='group',
          hover_data=['Details'],
          pattern_shape='Region',
          text_auto=True,
          color_discrete_map={'Eats':'red', 'Wst':'green'},
          color_discrete_sequence = ['yellow'],
          color='Region',
          )
fig.show()
```
