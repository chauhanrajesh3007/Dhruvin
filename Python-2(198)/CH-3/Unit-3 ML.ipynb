## PB-190
You've been given a dataset with apartment area and price information. There's a noticeable non-linear relationship between 
area and price. To address this you intend to categorize them into 'High', 'Medium', and 'Low' groups. Prices above $3,000,000 
are 'High', below $2,000,000 are 'Low', and between $2,000,000 and $3,000,000 are 'Medium'. Write a code to achieve this 
assuming that dataset has two columns named area and price

```python
import pandas as pd
import numpy as np
df=pd.DataFrame({'Area':[500,200,600,100,300],
               'Price':[3200000,1800000,3500000,1500000,2200000]})
df['Category']=np.where(df['Price']<2000000,'Low',
                   np.where(df['Price']>3000000,'High','Mid'))
print(df)
```

# PB-67
You are given a CSV file named ipl-matches.csv, which contains data of all Indian Premier League (IPL) matches played from the 
year 2008 to 2022.Read the dataset and display the basic information about the DataFrame.
Using Python and Pandas, write code to answer the following questions:
1. How many matches were won by Gujarat Titans in a Super Over ?
2. How Many Matches won by Royal Challengers Bangalore Vs Kolkata Knight Riders.
3. How many Matches won by Toss Winner of the Match.
4. In how many matches was the result decided by the D/L method and the match was won by wickets?
5. In how many matches Toss won by Rajasthan Royals & Won the Match.
6. In how many matches V Kohli declared player of the match against Chennai Super Kings.
7. Which Team is the Toss Winner of the Final match Played in Dubai?
8. List out the Finals played by Kolkata Knight Riders in which Toss Won by Kolkata Knight Riders in the Finals and choose to 
field.

```python
import pandas as pd
df=pd.read_csv('ipl-matches.csv')
df.info()
print('------------------------------------------------------------------------------')
print('1.How many matches were won by Gujarat Titans in a Super Over ?')
print(df.loc[(df['WinningTeam']=='Gujrat Titans')&(df['SuperOver']=='Y')].shape[0])
print('------------------------------------------------------------------------------')
print('2.How Many Matches won by Royal Challengers Bangalore Vs Kolkata Knight Riders.')
print(df.loc[(((df['Team1']=='Royal Challengers Bangalore')&(df['Team2']=='Kolkata Knight Riders'))|
    ((df['Team1']=='Kolkata Knight Riders')&(df['Team2']=='Royal Challengers Bangalore')))&
    (df['WinningTeam']=='Royal Challengers Bangalore')].shape[0])
print('------------------------------------------------------------------------------')
print('3.How many Matches won by Toss Winner of the Match.')
print(df.loc[(df['WinningTeam']==df['TossWinner'])].shape[0])
print('------------------------------------------------------------------------------')
print('4.In how many matches was the result decided by the D/L method and the match was won by wickets?')
print(df.loc[(df['method']=='D/L')&(df['WonBy']=='Wickets')].shape[0])
print('------------------------------------------------------------------------------')
print('5.In how many matches Toss won by Rajasthan Royals & Won the Match.')
print(df.loc[(df['TossWinner']=='Rajasthan Royals')&(df['WinningTeam']=='Rajasthan Royals')].shape[0])
print('------------------------------------------------------------------------------')
print('6.In how many matches V Kohli declared player of the match against Chennai Super Kings.')
print(df.loc[(df['Player_of_Match']=='V Kohli')&((df['Team1']=='Chennai Super Kings')|(df['Team2']=='Chennai Super Kings'))].shape[0])
print('------------------------------------------------------------------------------')
print('7.Which Team is the Toss Winner of the Final match Played in Dubai?')
print(df.loc[(df['City']=='Dubai')&(df['MatchNumber']=='Final')]['TossWinner'])
print('------------------------------------------------------------------------------')
print('8.List out the Finals played by Kolkata Knight Riders in which Toss Won by Kolkata Knight Riders in the Finals and choose to field.')
print(df.loc[(df['TossWinner']=='Kolkata Knight Riders')&(df['MatchNumber']=='Final')&(df['TossDecision']=='field')])
```
